```python
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

highest_employee = max(employees, key=employees.get)

print("Highest Salary:", highest_employee, "-", employees[highest_employee])

```

```python
numbers = [45, 22, 89, 10, 66]
maximum = numbers[0]
minimum = numbers[0]
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num
print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)
```

```python
prices = [450, 1200, 899, 1500, 300]
count = 0
for price in prices:
    if price > 1000:
        count += 1
print("Products above 1000:", count)
```

```python
attendance = ["P", "P", "A", "P", "P"]
present_count = attendance.count("P")
total_days = len(attendance)
attendance_percentage = (present_count / total_days) * 100
print("Attendance Percentage:", attendance_percentage)
```

```python

text = "pythonp"
char_count = {}
for char in text:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1
print(char_count)
```

```python
# List of student marks
marks = [70, 75, 80, 65]   # Example marks

# Initialize total
total = 0

# Calculate sum of marks using loop
for mark in marks:
    total += mark

# Calculate average
average_marks = total / len(marks)

# Display result
print("Average Marks:", average_marks)
```
