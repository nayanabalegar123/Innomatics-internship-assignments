```python
# Function to validate recharge amount and plan
def validate_recharge(amount, plan):
    valid_plans = [199, 299, 399, 599]

    # Check minimum recharge amount
    if amount < 50:
        print("❌ Recharge amount must be at least ₹50.")
        return False

    # Check if plan is valid
    if plan not in valid_plans:
        print("❌ Invalid plan. Available plans: 199, 299, 399, 599")
        return False

    return True


# Main program with retry option
while True:
    amount = int(input("Enter recharge amount: ₹"))
    plan = int(input("Enter plan (199, 299, 399, 599): "))

    if validate_recharge(amount, plan):
        print("✅ Recharge successful!")
        break
    else:
        print("Please try again.\n")
```

```python
# Function to check inventory stock
def check_inventory(products):
    for product, stock in products.items():
        if stock < 15:
            print(f"{product}: ⚠️ Reorder Alert (Stock = {stock})")
        else:
            print(f"{product}: ✅ Stock OK (Stock = {stock})")


# Dictionary of products and their stock quantities
inventory = {
    "Rice": 10,
    "Wheat": 25,
    "Sugar": 12,
    "Oil": 30,
    "Salt": 14
}

# Call function
check_inventory(inventory)
```

```python
# Function to process student result
def process_result(marks):
    total = 0

    # Calculate total using loop
    for mark in marks:
        total += mark

    # Calculate average
    average = total / len(marks)

    # Check pass or fail
    if average >= 50:
        return average, "Pass"
    else:
        return average, "Fail"


# List of student marks
student_marks = [60, 45, 70, 55, 40]

# Call function
avg, result = process_result(student_marks)

# Print result
print("Average Marks:", avg)
print("Result:", result)
```

```python
# Function to calculate cab fare
def calculate_fare(distance, peak):
    base_fare = 50
    rate_per_km = 12

    fare = base_fare + (distance * rate_per_km)

    # Add 25% extra for peak hour
    if peak.lower() == "yes":
        fare = fare + (fare * 0.25)

    return fare


# Main program with retry option
while True:
    distance = float(input("Enter distance travelled (in km): "))
    peak = input("Is it peak hour? (yes/no): ")

    total_fare = calculate_fare(distance, peak)

    print("Total Fare: ₹", total_fare)

    retry = input("Do you want to calculate again? (yes/no): ")
    if retry.lower() != "yes":
        print("Thank you for using Cab Fare Estimator!")
        break
```

```python
# Function to check attendance eligibility
def check_eligibility(attendance_list):
    present_count = 0

    # Count present days using loop
    for day in attendance_list:
        if day == "P":
            present_count += 1

    # Calculate attendance percentage
    percentage = (present_count / len(attendance_list)) * 100

    # Check eligibility
    if percentage >= 75:
        return percentage, "Eligible"
    else:
        return percentage, "Not Eligible"


# Employee attendance list
attendance = ["P", "P", "A", "P", "P", "A", "P", "P"]

# Call function
percent, result = check_eligibility(attendance)

# Print result
print("Attendance Percentage:", percent, "%")
print("Status:", result)
```

```python
# Function to check password strength
def check_password_strength(password):
    has_digit = False
    has_special = False
    special_chars = "@#$"

    # Check minimum length
    if len(password) < 8:
        return "Weak Password ❌ (Minimum length should be 8)"

    # Loop through password characters
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in special_chars:
            has_special = True

    # Check conditions
    if has_digit and has_special:
        return "Strong Password ✅"
    else:
        return "Weak Password ❌ (Must contain at least one digit and one special character @#$)"


# Take user input
password = input("Enter password: ")

# Check strength
result = check_password_strength(password)
print(result)
```
