```python
# ===============================
# 🚀 LANGCHAIN DEEP TECHNICAL BLOG (FINAL COLAB NOTEBOOK)
# ===============================

# =====================================
# 1. INSTALL DEPENDENCIES
# =====================================
!pip install -q transformers

# =====================================
# 2. IMPORT LIBRARIES
# =====================================
from transformers import pipeline
from IPython.display import Markdown, Image, display

# =====================================
# 3. LOAD MODEL (SAFE & WORKING)
# =====================================
pipe = pipeline("text-generation", model="gpt2")

# =====================================
# 4. GENERATION FUNCTION
# =====================================
def generate(text):
    prompt = f"Explain clearly in simple terms:\n{text}\nAnswer:"
    result = pipe(prompt, max_length=80, do_sample=True, temperature=0.7)
    return result[0]["generated_text"]

# =====================================
# 5. INTRODUCTION (BLOG SECTION)
# =====================================
display(Markdown("""
# 🧠 Introduction to LangChain

LangChain is a framework used to build applications powered by Large Language Models (LLMs).

## Why LangChain?
- Simplifies AI application development
- Enables chaining of LLM calls
- Connects tools, memory, and external APIs

## Problems it solves:
- Prompt complexity
- Multi-step reasoning
- Tool integration
"""))

# =====================================
# 6. BASIC LLM OUTPUT
# =====================================
print("🔹 Basic LLM Output:")
print(generate("What is LangChain?"))

# =====================================
# 7. PROMPT TEMPLATE SIMULATION
# =====================================
def prompt_template(topic, tone):
    return f"Explain {topic} in a {tone} tone"

print("\n🔹 Prompt Template Output:")
print(generate(prompt_template("Machine Learning", "simple")))

# =====================================
# 8. CHAIN SIMULATION
# =====================================
def chain(topic, tone):
    prompt = prompt_template(topic, tone)
    return generate(prompt)

print("\n🔹 Chain Output:")
print(chain("Artificial Intelligence", "beginner-friendly"))

# =====================================
# 9. MEMORY SIMULATION
# =====================================
chat_history = []

def chat(user_input):
    chat_history.append("User: " + user_input)
    context = " ".join(chat_history)
    response = generate(context)
    chat_history.append("AI: " + response)
    return response

print("\n🔹 Memory Example:")
print(chat("Hi, I am Nayana"))
print(chat("What is my name?"))

# =====================================
# 10. TOOL (CALCULATOR)
# =====================================
def calculator(expr):
    return eval(expr)

print("\n🔹 Calculator Tool:")
print("45 + 10 * 2 =", calculator("45 + 10 * 2"))

# =====================================
# 11. AGENT SIMULATION
# =====================================
def agent(query):
    if "calculate" in query.lower():
        expr = query.lower().replace("calculate", "")
        return calculator(expr)
    else:
        return generate(query)

print("\n🔹 Agent Output:")
print(agent("Calculate 25 * 4"))

# =====================================
# 12. DOCUMENT PROCESSING (SIMPLIFIED)
# =====================================
text = "LangChain helps build applications using LLMs."

chunks = [text[i:i+25] for i in range(0, len(text), 25)]

print("\n🔹 Document Chunks:")
print(chunks)

# =====================================
# 13. ARCHITECTURE DIAGRAM
# =====================================
display(Markdown("""
# 🏗️ Architecture Flow

User Input → Prompt → LLM → Chain → Tool/Agent → Output
"""))

# Safe working image
display(Image(url="https://raw.githubusercontent.com/langchain-ai/langchain/master/docs/static/img/langchain_architecture.png"))

# =====================================
# 14. REAL-WORLD USE CASES
# =====================================
display(Markdown("""
# 🌍 Real-World Use Cases

## 1. AI Chatbots
Uses memory + LLM

## 2. Document Q&A Systems
Uses loaders + vector DB

## 3. AI Agents
Uses tools + reasoning systems
"""))

# =====================================
# 15. ADVANTAGES & LIMITATIONS
# =====================================
display(Markdown("""
# ✅ Advantages
- Modular architecture
- Easy to build AI apps
- Flexible integrations

# ❌ Limitations
- Slower for large models
- Debugging complexity
- Token cost in APIs

# 🚫 When NOT to use
- Simple scripts
- Small offline tasks
"""))

# =====================================
# 16. CONCLUSION
# =====================================
display(Markdown("""
# 🎯 Conclusion

LangChain is a powerful framework for building AI applications.

## Future Scope:
- LangGraph
- Multi-agent AI systems
- Autonomous AI workflows
"""))

# =====================================
# ⭐ END NOTEBOOK
# =====================================
print("\n✅ Notebook executed successfully!")
```
