```python
# Function to analyze post engagement
def analyze_engagement(likes_list):
    total_likes = 0

    # Loop to calculate total likes
    for likes in likes_list:
        total_likes += likes

    # Check post status
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    # Output
    print("Total Likes:", total_likes)
    print("Post Status:", status)


# Example usage
likes = [200, 300, 150, 400]   # sample likes from different sources
analyze_engagement(likes)
```

```python
# Function to monitor medicine stock
def check_medicine_stock(stock):

    # Check stock status using conditional logic
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"

    # Output
    print("Medicine Stock:", stock)
    print("Status:", status)


# Example usage
stock = 6   # sample stock value
check_medicine_stock(stock)
```

```python
# Function to check rainfall adequacy
def check_rainfall(rainfall_data, required_level):

    # Calculate total rainfall using loop
    total = 0
    for rain in rainfall_data:
        total += rain

    # Calculate average rainfall
    average = total / len(rainfall_data)

    # Check rainfall status
    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    # Output
    print("Average Rainfall:", int(average))
    print("Rainfall Status:", status)


# Example usage (seasonal rainfall values)
rainfall = [70, 75, 68, 80, 67]   # rainfall data
required = 70                      # required level

check_rainfall(rainfall, required)
```

```python
# Function to detect duplicate usernames
def detect_duplicates(usernames):

    unique_users = set(usernames)

    # Check if duplicates exist
    if len(unique_users) != len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


# Example usage
accounts = ["john123", "alice22", "john123", "mike09"]

detect_duplicates(accounts)
```

```python
# Function to filter premium crop prices
def premium_crop_filter(prices):
    premium = []

    # Loop through prices
    for price in prices:
        if price > 2000:
            premium.append(price)

    return premium


# Example crop prices
crop_prices = [1500, 2500, 1800, 3200, 900]

# Get premium crops
premium_crops = premium_crop_filter(crop_prices)

# Output
print("Premium Crops:", premium_crops)
```

```python
# Function to check application health
def application_health_checker(errors):
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print("Error Count:", errors)
    print("System Status:", status)


# Example input
error_count = 7

# Call function
application_health_checker(error_count)
```

```python
# Function to check daily transaction limit
def transaction_limit_checker(amount):
    daily_limit = 50000

    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"

    print("Transaction Amount:", amount)
    print("Transaction Status:", status)


# Example input
transaction_amount = 60000

# Call function
transaction_limit_checker(transaction_amount)
```

```python
# Function to calculate attendance and check eligibility
def attendance_eligibility(attendance_list):
    total_classes = len(attendance_list)
    attended_classes = 0

    # Count attended classes
    for status in attendance_list:
        if status == 1:   # 1 = Present, 0 = Absent
            attended_classes += 1

    # Calculate attendance percentage
    percentage = (attended_classes / total_classes) * 100

    # Check eligibility
    if percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"

    print("Attendance Percentage:", percentage)
    print("Exam Eligibility:", eligibility)


# Example input (1 = Present, 0 = Absent)
attendance_data = [1, 1, 1, 1, 0]   # 4 out of 5 classes attended → 80%

# Call function
attendance_eligibility(attendance_data)
```
