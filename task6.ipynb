```python
def smart_parking(capacity, logs):
    current_vehicles = 0
    peak_usage = 0

    # Process each log
    for log in logs:
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT":
            if current_vehicles > 0:
                current_vehicles -= 1

        # Track peak usage
        if current_vehicles > peak_usage:
            peak_usage = current_vehicles

    # Print current parked vehicles
    print("Currently Parked Vehicles:", current_vehicles)

    # Check capacity
    if current_vehicles > capacity:
        print("Parking Status: Exceeded Capacity! Alert!")
    else:
        print("Parking Status: Available")

    # Optional: show peak usage
    print("Peak Parking Usage:", peak_usage)


# Sample Input
capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking(capacity, logs)
```

```python
def estimate_delivery_time(distance, traffic, weather):
    # Base time calculation (Assume 5 minutes per km)
    eta = distance * 5

    # Traffic delay
    if traffic == "High":
        eta += 15
    elif traffic == "Medium":
        eta += 10
    elif traffic == "Low":
        eta += 5

    # Weather delay
    if weather == "Rainy":
        eta += 10
    elif weather == "Stormy":
        eta += 20
    elif weather == "Clear":
        eta += 0

    print("Estimated Delivery Time:", eta, "minutes")


# Sample Input
distance = 8
traffic = "High"
weather = "Rainy"

estimate_delivery_time(distance, traffic, weather)
```

```python
def analyze_occupancy(total_seats, booked_seats):
    # Count booked seats
    booked_count = 0
    for seat in booked_seats:
        if seat == 1:
            booked_count += 1

    # Calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100

    print("Occupancy:", int(occupancy), "%")

    # Determine show status
    if occupancy == 100:
        print("Show Status: Housefull")
        print("Suggestion: Open Additional Show")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")


# Sample Input
total_seats = 200
booked_seats = [1] * 150   # 150 booked seats

analyze_occupancy(total_seats, booked_seats)
```

```python
def classify_server_load(cpu_readings):
    total = 0

    # Calculate total CPU usage
    for reading in cpu_readings:
        total += reading

    # Calculate average
    average = total / len(cpu_readings)

    print("Average CPU Load:", int(average), "%")

    # Classify load
    if average < 50:
        print("Server Status: Normal")
    elif 50 <= average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# Sample Input
cpu_readings = [45, 60, 70, 85, 90]

classify_server_load(cpu_readings)
```

```python
def resource_monitor(resource_usage):
    overused = []

    # Define usage limit (for example: 8 hours per day)
    limit = 8

    # Check each resource
    for resource, hours in resource_usage.items():
        if hours > limit:
            overused.append(resource)

    # Display results
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


# Sample Input
usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

resource_monitor(usage)
```

```python
def manage_registrations(capacity, registrations):
    confirmed = 0
    waitlisted = 0

    # Process each registration
    for i in range(registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1

    # Determine registration status
    if confirmed >= capacity:
        status = "Closed"
    else:
        status = "Open"

    # Display results
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


# Sample Input
capacity = 100
registrations = 105

manage_registrations(capacity, registrations)
```
